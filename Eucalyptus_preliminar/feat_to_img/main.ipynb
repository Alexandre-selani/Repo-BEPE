```python
import pandas as pd
import numpy as np
import PIL
import os
from pathlib import Path

index_dataset1 = pd.read_csv("dataset20.csv")
index_dataset1 = index_dataset1.iloc[:, 1:]
index_dataset1.columns = ["y","x"]

index_geral = pd.read_csv("index_geral.csv")
index_geral = index_geral.iloc[:, 1:]
index_geral.columns.values[0] = "y"
index_geral.columns.values[1] = "x"

for i,column in enumerate(index_geral.columns.values[2:-1]):
    index_geral.columns.values[i+2] = f"feature{i+1}"
    print(column)

index_geral.columns.values[-1] = "class"

index_geral = index_geral.merge(index_dataset1,on=["y","x"])

index_geral.to_csv("../hc-features/dataset20_index_features.csv",index=False)
```
